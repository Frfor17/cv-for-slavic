```python
from ultralytics import YOLO

model = YOLO("yolov8n.pt")  # or yolov8s.pt
```

```python
model.train(
    data="data.yaml",
    imgsz=640,
    epochs=10,      # да, 10 эпох на 3 картинки — это уже “много”
    batch=2,        # 2‑3 картинки в батче
    name="box_det"
)
```

```python
results = model("images/image1.jpg")
results.show()
```
