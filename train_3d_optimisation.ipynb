```python
# cell 3: fake 3D boxes from YOLO + manual calibration
# Пример: после детекции YOLO ты вручную заполнил реальные размеры
boxes_3d = [
    {"id": 0, "l": 0.3,  "w": 0.2,  "h": 0.1},
    {"id": 1, "l": 0.25, "w": 0.15, "h": 0.15},
    {"id": 2, "l": 0.2,  "w": 0.3,  "h": 0.1},
]

shelf = {"l": 1.0, "w": 0.4, "h": 0.5}

print("Boxes 3D:", boxes_3d)
print("Shelf:", shelf)
```

```python
# cell 4: 3D bin packing
from py3dbp import Packer, Bin, Item

packer = Packer()

# 1. Полка = один bin
bin_name = "shelf"
packer.add_bin(Bin(
    name=bin_name,
    width=shelf["w"],   # ширина полки
    height=shelf["h"],  # высота полки
    depth=shelf["l"],   # глубина/длина полки
    max_weight=100.0    # любой большой вес
))

# 2. Добавляем коробки как Item
for b in boxes_3d:
    # l, w, h → depth, width, height в py3dbp
    packer.add_item(Item(
        name=f"box_{b['id']}",
        width=b["w"],
        height=b["h"],
        depth=b["l"],
        weight=1.0
    ))

# 3. Запуск упаковки
packer.pack()

# 4. Вывод результата
for b in packer.bins:
    print(b.string())
    print("FITTED ITEMS:")
    for item in b.items:
        print("  ", item.string())
    print("UNFITTED ITEMS:")
    for item in b.unfitted_items:
        print("  ", item.string())
```
